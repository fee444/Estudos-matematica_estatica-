Exercício 3: Otimização de Produto EscalarDescrição: Encontre o vetor unitário que maximiza o produto escalar com o vetor v = [3, 4, 5]. Use técnicas de otimização para encontrar a solução.

```python
from scipy.optimize import minimize

# ARRAY
v = np.array([3, 4, 5])

#Função Objetivo: Queremos pegar o valor negativo do produto escalar entre o nosso vetor inicial(v) e o x que será calculado

def funcao_objetivo(x):
    return -np.dot(v,x)
    
# Restrição para gerar um vetor unitário
constraint = {'type' : 'eq', 'fun' : lambda x: np.linalg.norm(x) - 1}

# Solução de otimização

resultado = minimize(funcao_objetivo, np.ones(3), constraints = constraint)
vetor_otimizado = resultado.x

print(vetor_otimizado)


# Calculando o vetor otimizado

produto_escalar = np.dot(v, vetor_otimizado)
print(float(produto_escalar))
```

Exercício 4: Produto Vetorial e Direção NormalDescrição:  Dados  três  pontos  no  espaço  3D  A  =  [1,  2,  3],  B  =  [4,  5,  6],  e  C  =  [7,  8,  9], encontre um vetor normal ao plano definido por esses três pontos

```python
A  =  np.array([1,  2,  3])
B  =  np.array([4,  5,  6])
C  =  np.array([7,  8,  9])

AC = C - A 
AB = B - A 

resultado = np.cross(AC, AB)

print(resultado)
```

VIZUALIZAÇÕES DE SUBESPAÇO (R2, R3 E R4)

```python

fig, ax = plt.subplots(figsize=(8, 8))

# Valores de X e Y
x = np.arange(-10, 11, 1)
y = 4/6 * x

# Plot da reta que passa pela origem
ax.plot(
    x,
    y,
    lw=4,
    color='green',
    label=r'$y=\frac{2}{3}x$, subespaço de $\mathbf{R}^2$'
)

# Novo valor de Y
y = -3 + 4/6 * x

# Plot da reta que não passa pela origem
ax.plot(
    x,
    y,
    lw=4,
    color='red',
    label=r'$y=-3+\frac{2}{3}x$, não é um subespaço de $\mathbf{R}^2$'
)

# Grid
ax.grid(True)

# Título
ax.set_title('Visualização de Subespaço em $R^2$', size=18)

# Ponto na origem
ax.scatter(0, 0, s=100, fc='black', ec='red')

# Texto e legenda
ax.text(-2, 0, '$(0,0)$', size=18)
ax.legend(fontsize=12)

# Coordenadas dos eixos
ax.axis([-10, 10, -10, 10])

plt.show()
```
